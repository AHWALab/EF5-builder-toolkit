# Step 2: Download precipitation

Run from `Example/Example_notebooks` with the **EF5 Builder** kernel.
Choose the product you need and run its example. Set the UTC period in each example. Files are saved
in separate folders under `Example/data_model/precip`.

| Product | Time step | Units |
| --- | --- | --- |
| IMERG Final or Early | 30 minutes | mm/h |
| MRMS Rate | 2 minutes | mm/h |
| MRMS Pass2 | 1 hour | mm |

IMERG needs a [PPS-registered email](https://registration.pps.eosdis.nasa.gov/).
The MRMS examples use a U.S. area. To inspect files you already downloaded,
run the setup and inspection cells. [Product details](../../docs/precipitation.md).


```python
import logging
from pathlib import Path
import rasterio
from ef5_builder import download_precipitation

logging.basicConfig(level=logging.INFO)
precip_dir = Path("../data_model/precip")
product_folders = ("imerg_final", "imerg_early", "mrms_rate", "mrms_pass2")
for name in product_folders:
    (precip_dir / name).mkdir(parents=True, exist_ok=True)

email = "vrobledodelgado@uiowa.edu"  # Replace with your PPS-registered email.
overwrite = True  # Set True when you intentionally rerun these examples.

# IMERG outputs cover the Ghana example DEM extent.
with rasterio.open("../data_model/basic/dem.tif") as dem:
    if dem.crs.to_epsg() != 4326:
        raise ValueError("The example DEM must be in EPSG:4326.")
    imerg_bounds = tuple(dem.bounds)
mrms_bounds = (-95.0, 40.0, -93.0, 42.0)
downloads = {}
print("Output folder:", precip_dir.resolve())
```

## 1. IMERG Final

This example downloads six hours at 30-minute intervals. Filenames show the
end of each interval, so the 00:00 source becomes `imerg.202506010030.tif`.


```python
downloads["imerg_final"] = download_precipitation(
    product="IMERG",
    imerg_run="final",
    start="2025-06-01 00:00:00",
    end="2025-06-01 06:00:00",
    output_dir=precip_dir / "imerg_final",
    email=email,
    bounds=imerg_bounds,
    overwrite=overwrite,
)
print("IMERG Final files:", len(downloads["imerg_final"]))
downloads["imerg_final"]
```

## 2. IMERG Early

Download one hour using Early Run. Keep Early and Final in
separate folders because their output filenames use the same pattern.


```python
downloads["imerg_early"] = download_precipitation(
    product="IMERG",
    imerg_run="early",
    start="2025-06-01 00:00:00",
    end="2025-06-01 01:00:00",
    output_dir=precip_dir / "imerg_early",
    email=email,
    bounds=imerg_bounds,
    overwrite=overwrite,
)
print("IMERG Early files:", len(downloads["imerg_early"]))
downloads["imerg_early"]
```

## 3. MRMS Rate

Download precipitation rates for the U.S. example area. One hour contains
30 files. The required GRIB driver is included in `environment.yml`.
[Driver troubleshooting](../../docs/precipitation.md#mrms-grib-dependency).


```python
downloads["mrms_rate"] = download_precipitation(
    product="MRMS",
    mrms_product="Rate",
    start="2025-06-01 00:00:00",
    end="2025-06-01 01:00:00",
    output_dir=precip_dir / "mrms_rate",
    bounds=mrms_bounds,
    overwrite=overwrite,
)
print("MRMS Rate files:", len(downloads["mrms_rate"]))
downloads["mrms_rate"][:3]
```

## 4. MRMS Pass2

Download one hourly total in **mm**.


```python
downloads["mrms_pass2"] = download_precipitation(
    product="MRMS",
    mrms_product="Pass2",
    start="2025-06-01 00:00:00",
    end="2025-06-01 01:00:00",
    output_dir=precip_dir / "mrms_pass2",
    bounds=mrms_bounds,
    overwrite=overwrite,
)
print("MRMS Pass2 files:", len(downloads["mrms_pass2"]))
downloads["mrms_pass2"]
```

## Check the downloaded outputs

Inspect the files, units, and timestamps in each product folder. This cell
also works with files from an earlier run.


```python
for name in product_folders:
    files = sorted((precip_dir / name).glob("*.tif"))
    print(f"\n{name}: {len(files)} files")
    if not files:
        continue
    with rasterio.open(files[0]) as src:
        print("First file:", files[0].name)
        print("Shape:", src.shape, "CRS:", src.crs, "NoData:", src.nodata)
        print("Units:", src.tags().get("units"), "UTC time:", src.tags().get("timestamp_utc"))
        if src.profile.get("tiled", False) or src.compression != rasterio.enums.Compression.deflate:
            raise ValueError(f"Expected striped, Deflate-compressed output: {files[0]}")
```

Use UTC dates aligned to the product's time step. `start` is included and
`end` is excluded for source times. Set `overwrite=True` to replace existing files.
See [README Step 2](../../README_library.md#step-2-prepare-precipitation-forcing-data)
for the EF5 forcing block and [technical details](../../docs/precipitation.md)
for naming and unit conversions.
